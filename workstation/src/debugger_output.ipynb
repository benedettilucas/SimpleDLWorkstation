```python
from importlib.metadata import distribution, PackageNotFoundError

instalados = []
ausentes = []

for pacote in pacotes:
    try:
        distribution(pacote)
        instalados.append(pacote)
    except PackageNotFoundError:
        ausentes.append(pacote)

print(f"Instalados ({len(instalados)}):")
print(", ".join(instalados) if instalados else "Nenhum")

print(f"\nNão instalados ({len(ausentes)}):")
print(", ".join(ausentes) if ausentes else "Nenhum")
```
