```python
pacotes = [
    "numpy",
    "pandas",
    "scipy",
    "scikit-learn",
    "matplotlib",
    "einops",
    "pytorch-lightning",
    "tensorboard",
    "torchmetrics",
    "opencv-python",
    "imageio",
    "scikit-image",
    "pywavelets",
    "transformers",
    "huggingface-hub",
    "timm",
    "omegaconf",
    "segment-anything",
    "pretrainedmodels",
    "grad-cam",
    "tqdm",
    "requests",
    "protobuf",
    "six",
    "hydra-core",
    "sentencepiece",
    "torch",
    "torchvision",
    "torchaudio",
]
```

```python
from importlib.metadata import distribution, PackageNotFoundError

instalados = []
ausentes = []

for pacote in pacotes:
    try:
        distribution(pacote)
        instalados.append(pacote)
    except PackageNotFoundError:
        ausentes.append(pacote)

print(f"Instalados ({len(instalados)}):")
print(", ".join(instalados) if instalados else "Nenhum")

print(f"\nNão instalados ({len(ausentes)}):")
print(", ".join(ausentes) if ausentes else "Nenhum")
```
